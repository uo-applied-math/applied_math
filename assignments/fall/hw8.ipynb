# (1) Wilson's Algorithm - random spanning tree

The following is Wilson's algorithm, for finding a random spanning tree.  It was designed by David Wilson, and first written up [here](https://dl.acm.org/doi/pdf/10.1145/237814.237880).  It uses the loop erased random walk from the last homework.

Pick a root, $r$, and let $T$ be the single-vertex tree $\{r\}$.  Then, repeat the following:
- Choose a vertex v that is not in $T$.
- perform loop erased random walk, as in the above question, starting at $v$ until we arrive at a vertex $t$ in $T$.  Add the walk to $T$.

Then, forget the root.  $T$ is now a uniformly random spanning tree.  We'll prove this in class.

(b) Test Wilson's algorithm by running it many times on the $2 \times 3$ grid.  This graph doesn't have too many spanning trees.  Produce a histogram which shows that you're getting the uniform distribution on the spanning trees.

(c) Figure out how big you can make $n$ and still get an $n \times n$ square-grid spanning tree in a reasonable amount of time on your computer.

# (3) Ising model

References:
- Wikipedia page for the [Ising model](https://en.wikipedia.org/wiki/Ising_model)
- David Levin's book, [Markov Chains and Mixing Times](https://pages.uoregon.edu/dlevin/MARKOV/mcmt2e.pdf), Chapters 1 (Markov chain basics), 3 (Glauber dynamics), 15 (Ising model).  In the book, the goal is in proving theorems about mixing and cutoff for these chains; we don't really have time to do anything serious along those lines.

In this final sequence of problems we consider the square-lattice Ising model, which is a famous toy model for a ferromagnet.  This is going to be a special case of the model on the Wikipedia and in David's book, with a bunch of constants set to zero or 1.

An *Ising spin* is an element of the set $\{+1, -1\}$.  The lattice $\Lambda$ is the square grid $\{0, \dots, n-1\} \times \{0, \dots, n-1\}$. An *Ising state* is a function $\sigma: \Lambda \rightarrow \{\pm 1\}$ which assigns a spin to each lattice vertex; we sometimes write $\sigma_j$ instead of $\sigma(j)$.  So, there are $2^{n^2}$ ising states.

The *Hamiltonian*, or *energy* of an Ising state $\sigma$ is then

$$H(\sigma) = -\sum_{i \sim j} \sigma_i \sigma_j$$

where the sum is over all pairs of adjacent vertices in $\Lambda$.  Fix the constant $\beta > 0$, called the *inverse temperature*, and define the *partition function* by

$$Z_{\beta} = \sum_{\sigma} e^{-\beta H(\sigma)},$$

(so, yes, the two minus signs cancel out; they're there for convention reasons).  Finally define the probability measure 

$$P(\sigma) = \frac{1}{Z_{\beta}}e^{-\beta H(\sigma)}$$

on Ising states. 

In this problem and the next, we'll sample a random configuration from the Ising model.  In this first problem, we'll do it carelessly; in the second, carefully.  For both we'll consider [Glauber, or "heat bath", dynamics](https://en.wikipedia.org/wiki/Glauber_dynamics) for the Ising model.  This is the markov chain which selects $i \in \Lambda$ uniformly at random, and then replaces $\sigma_{x,y}$ with a $+1$ with probability 
$$ p(\Delta E) = \frac{1}{1+e^{\delta \Delta E}}$$

and -1 with probability $1-p(\Delta E)$, where

$$ \Delta E = 2 \sigma_{x,y} (\sigma_{x+1,y} + \sigma_{x-1, y} + \sigma_{x, y+1} + \sigma_{x, y-1}) $$

(a) Explain why running this markov chain for long enough leads to a sample from the probability measure $P$.

(b) Without worrying too much about what "long enough" is, implement the markov chain, so as to draw samples from a $20 \times 20$ grid $\Lambda$ (Analyzing the *mixing time* of markov chains is a classical topic, and also a research topic, in probability and in statistical mechanics).  If you can do a bigger grid than $20 \times 20$, please do.  If you have to decrease the size of the grid, then also that is OK, though if it's too small it might be hard to draw conclusions.

(c) What does the value of $\beta$ control in this model? If you're not sure, try running the chain for longer.  In particular, Wikipedia says there's a "critical" value, around $\beta = 2.2$, at which point the behavior of the model abruptly changes.  What happens at the critical value? Is wikipedia correct about what this critical value is?  Draw pictures of your samples as bitmaps (-1 = black pixel, 1 = white pixel), with $\beta$ below, at, and above the critical value.  


# (4) Coupling from the past

- Reference: David Levin's book [Markov Chains and Mixing Times](https://pages.uoregon.edu/dlevin/MARKOV/mcmt2e.pdf), Chapter 25, sections 1-3 (and preceding sections if you don't know some of the terminology).

We continue studying Glauber Dynamics on the Ising model, but this time, we'll use Monotone coupling from the past, (MCFTP), instead of not caring about the mixing time.  This is written up in the reference, so read that first. I'll also discuss this in class in Week 10. 

Ising states have a partial order: product order, which is a ranked poset (the rank is the number of plus signs).  The minimal state is $\hat{0}$ (all -1 spins) and the maximal state is $\hat{1}$ (all +1 spins).  Quoting from David's book:

```
Recall our supposition that the randomizing process, in this case the single-
site heat bath, has been running for all time. Suppose that someone has recorded
all the randomizing operations of the heat bath up until the present time. They
have not recorded what the actual spin configurations or Markov chain transitions
are, but merely which sites were updated and which random number was used to
update the spin at the given site. Given this recorded information, our goal is to
determine the state of the Markov chain at the present time (time 0), since, as we
have already determined, this state is a sample from the stationary distribution of
the Markov chain.
```

So we simulate the history of the Markov chain - if our simulation "collapses" the minimal state $\hat{0}$ and the maximal state $\hat{1}$ into the same state by the time the simulation completes, then the final state of the simulation is randomly distributed according to the stationary measure of the markov chain.  Please read more of the reference for more details; part (b) asks you explicitly to do this.

(a) Implement MCFTP as follows.  Let $N = 100$ or something.  Then: 
- Make a list of "randomizing" instructions of length N: each instruction is of the form $(i, j, q)$ where $(i,j) \in L$ and $q$ is a uniformly random number between 0 and 1.
- apply the list of randomizing instructions to the minimal state $\hat{0}$: for each $(i,j,q)$, resample state $(i,j)$ as you did in problem 4 (the new spin should be +1 if $q < p(\Delta E)$ and -1 otherwise.  This will produce a state $s$.
- do the same for $\hat{1}$, producing a state $t$.
- Claim: $s \leq t$ in the partial order.  Compute the ranks of $s$ and $t$.
  - If the ranks are the same, then actually $s=t$, so stop; you've sampled a random Ising state.
  - If the ranks aren't the same, then we haven't simulated enough of the history of the chain.  Make another list of $N$ "randomizing" instructions, *prepend* it to your current list, and double N.  Then repeat these steps.
  
Note: you have to do exactly these steps.  Most small variations of this procedure are wrong, and introduce subtle bias.  In particular: always follow all $N$ instructions - don't stop the simulation early.
